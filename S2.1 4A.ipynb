```python
def pz (radius,x_centre,y_centre,orientation):
    angle=45
    seg=sin(angle*pi/180/2)*radius*2
    y_change=cos(angle*pi/180/2)*seg
    x_change=(seg**2-y_change**2)**0.5

    right=x_centre+radius-x_change
    left=x_centre-radius+x_change

    if orientation=='up':
        new_y=y_centre+radius*1.5-y_change
        test_arc=Arc((x_centre,y_centre+radius*1.5),radius*2,radius*2,theta1=0,theta2=180+angle*2,angle=-angle,color='black')
        gcf().gca().add_artist(test_arc)
        plot([x_centre,right],[y_centre,new_y],color='black',linewidth=1)
        plot([x_centre,left],[y_centre,new_y],color='black',linewidth=1)

        new_y=y_centre-radius*1.5+y_change   
        test_circ2=Circle((x_centre,y_centre-radius*1.5),radius,color='black')
        gcf().gca().add_artist(test_circ2)
        x_coord=[x_centre,left,right]
        y_coord=[y_centre,new_y,new_y]
        fill(x_coord,y_coord,color='black')
        plot([x_centre,right],[y_centre,new_y],color='black',linewidth=1.5)
        plot([x_centre,left],[y_centre,new_y],color='black',linewidth=1.5) 

    elif orientation=='down':
        new_y=y_centre-radius*1.5+y_change
        test_arc=Arc((x_centre,y_centre-radius*1.5),radius*2,radius*2,theta1=0,theta2=180+angle*2,angle=180-angle,color='black')
        gcf().gca().add_artist(test_arc)
        plot([x_centre,right],[y_centre,new_y],color='black',linewidth=1)
        plot([x_centre,left],[y_centre,new_y],color='black',linewidth=1)

        new_y=y_centre+radius*1.5-y_change   
        test_circ2=Circle((x_centre,y_centre+radius*1.5),radius,color='black')
        gcf().gca().add_artist(test_circ2)
        x_coord=[x_centre,left,right]
        y_coord=[y_centre,new_y,new_y]
        fill(x_coord,y_coord,color='black')
        plot([x_centre,right],[y_centre,new_y],color='black',linewidth=1.5)
        plot([x_centre,left],[y_centre,new_y],color='black',linewidth=1.5) 
```

```python
def local_min_find(data,check_no,threshold,min_wave,max_wave,x_off,y_off):
    for i in range(len(data[:,1])):
        if i>check_no and i<len(data[:,1])-check_no:
            if data[i,1]<data[i-check_no,1] and data[i,1]<data[i+check_no,1] and data[i,1]<threshold and data[i,0]>min_wave and data[i,0]<max_wave:
                plot([data[i,0],data[i,0]],[data[i,1]-3,data[i,1]-8],linewidth=1,color='black')
                text(data[i,0]+x_off,data[i,1]-y_off,int(data[i,0]),fontsize=8, rotation=90)
       
```

```python
plot(ir_data[:,0],ir_data[:,1],linewidth=1)

xlim(4000,500)
ylim(20,)

title('IR spectrum of ICl.pyridine')
xlabel('Wavenumber cm$^{-1}$')
ylabel('% Transmitance')

ax=gca()
ax.xaxis.set_ticks_position('top')
ax.yaxis.set_ticks_position('right')
ax.xaxis.set_label_position('top')
ax.yaxis.set_label_position('right')

ax.spines['bottom'].set_color('none')
ax.spines['left'].set_color('none')

local_min_find(ir_data,1,65,1250,2000,20,14)
local_min_find(ir_data,1,77,2500,4000,25,14)
local_min_find(ir_data,1,77.5,2800,2950,25,14)

text(3110,58,'{',fontsize=25,rotation=90)
text(3000,40,'Aromatic C—H,',fontsize=8,rotation=90)
text(1605,30,'Aromatic C—C,',fontsize=8,rotation=90)
text(1460,18,'Aromatic C—C,',fontsize=8,rotation=90)   
text(640, 17, 'N—I stretch,',fontsize=8, rotation=90)

local_min_find(ir_data,1,65,650,1000,30,13)
local_min_find(ir_data,1,65,500,650,10,13)
local_min_find(ir_data,1,65,1000,1050,10,14)
local_min_find(ir_data,1,65,1050,1100,45,14)


savefig('IR ICl pyridine.jpeg',bbox_inches='tight',dpi=300)

show()
```

```python
def empty_p (angle,radius,x_centre,y_centre,orientation):
    seg=sin(angle*pi/180/2)*radius*2
    y_change=cos(angle*pi/180/2)*seg
    x_change=(seg**2-y_change**2)**0.5

    right=x_centre+radius-x_change
    left=x_centre-radius+x_change

    if orientation=='top':
        start_y=y_centre-radius*1.5
        new_y=y_centre-y_change
        test_arc=Arc((x_centre,y_centre),radius*2,radius*2,theta1=0,theta2=180+angle*2,angle=-angle,color='black')

    elif orientation=='bottom':
        start_y=y_centre+radius*1.5
        new_y=y_centre+y_change        
        test_arc=Arc((x_centre,y_centre),radius*2,radius*2,theta1=0,theta2=180+angle*2,angle=180-angle,color='black')        

    plot([x_centre,right],[start_y,new_y],color='black',linewidth=1)
    plot([x_centre,left],[start_y,new_y],color='black',linewidth=1)
    gcf().gca().add_artist(test_arc)

def filled_p (angle,radius,x_centre,y_centre,orientation):
    seg=sin(angle*pi/180/2)*radius*2
    y_change=cos(angle*pi/180/2)*seg
    x_change=(seg**2-y_change**2)**0.5

    right=x_centre+radius-x_change
    left=x_centre-radius+x_change

    if orientation=='top':
        start_y=y_centre-radius*1.5
        new_y=y_centre-y_change

    elif orientation=='bottom':
        start_y=y_centre+radius*1.5
        new_y=y_centre+y_change   
        
    test_circ2=Circle((x_centre,y_centre),radius,color='black')
    gcf().gca().add_artist(test_circ2) 

    x_coord=[x_centre,left,right]
    y_coord=[start_y,new_y,new_y]

    fill(x_coord,y_coord,color='black')
    plot([x_centre,right],[start_y,new_y],color='black',linewidth=1.5)
    plot([x_centre,left],[start_y,new_y],color='black',linewidth=1.5) 
```
